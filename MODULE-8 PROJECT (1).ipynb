### Problem Statement (Short)
Create a simple application that lets users apply basic image filters—like grayscale, blur, sepia, and edge detection—to any uploaded image. The goal is to make image editing quick, easy, and accessible without requiring advanced skills.

```python
import cv2
import numpy as np
```

```python
# Load the image from file
image = cv2.imread(r"C:\Users\NSTI\Downloads\car.jpg")
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
blur = cv2.GaussianBlur(image, (15, 15), 0)
```

```python
# Create sepia filter matrix
sepia_filter = np.array([[0.272, 0.534, 0.131],
                         [0.349, 0.686, 0.168],
                         [0.393, 0.769, 0.189]])

# Apply the filter
sepia = cv2.transform(image, sepia_filter)

# Clip values to be in valid range
sepia = np.clip(sepia, 0, 255).astype(np.uint8)
```

```python
edges = cv2.Canny(image, 100, 200)
```

```python
cv2.imwrite('output_gray.jpg', gray)
cv2.imwrite('output_blur.jpg', blur)
cv2.imwrite('output_sepia.jpg', sepia)
cv2.imwrite('output_edges.jpg', edges)
```

```python
cv2.imshow('Original', image)
cv2.imshow('Grayscale', gray)
cv2.imshow('Blur', blur)
cv2.imshow('Sepia', sepia)
cv2.imshow('Edges', edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
