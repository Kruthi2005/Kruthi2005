```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="12345",
    database="hospital_db"
)
cursor = conn.cursor()
```

```python
# Add doctor
def add_doctor(name, specialization):
    cursor.execute("INSERT INTO doctors (name, specialization) VALUES (%s, %s)", (name, specialization))
    conn.commit()

# Add patient
def add_patient(name, age):
    cursor.execute("INSERT INTO patients (name, age) VALUES (%s, %s)", (name, age))
    conn.commit()
```

```python
def book_appointment(patient_id, doctor_id, date):
    cursor.execute("INSERT INTO appointments (patient_id, doctor_id, appointment_date) VALUES (%s, %s, %s)",
                   (patient_id, doctor_id, date))
    conn.commit()
```

```python
def view_appointments():
    cursor.execute("""
        SELECT a.appointment_id, p.name AS patient, d.name AS doctor, a.appointment_date
        FROM appointments a
        JOIN patients p ON a.patient_id = p.patient_id
        JOIN doctors d ON a.doctor_id = d.doctor_id
    """)
    for row in cursor.fetchall():
        print(row)
```

```python
add_doctor("Dr. Smith", "Cardiology")
add_patient("Alice", 29)
book_appointment(1, 1, "2025-07-01")
view_appointments()
```
